```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Configuración opcional para que los gráficos de Plotly se rendericen correctamente en VS Code
pio.renderers.default = "vscode"

# Cargar el archivo subiendo un nivel en el directorio para llegar a la raíz
celulares_2024_df = pd.read_csv('../celulares_2024.csv')

# Visualizar las primeras 5 filas del conjunto de datos
celulares_2024_df.head(10)
```

```python
# Creamos una lista con los nombres exactos de las columnas que queremos conservar
columnas_utiles = [
    'phone_brand', 
    'phone_model', 
    'price_usd', 
    'storage', 
    'ram',
    'display_type',
    'usb',
    'battery',
    'colors',
    'video',
    'chipset',
    'cpu',
    'gpu',
    'os_type',
]

# Filtramos el DataFrame original y guardamos el resultado en un nuevo DataFrame limpio
df_limpio = celulares_2024_df[columnas_utiles]

# Verificamos la estructura del nuevo DataFrame resumido
df_limpio.info()
```

```python
# Mostramos las primeras 5 filas del DataFrame filtrado
print("--- VISTAZO A LOS DATOS FILTRADOS ---")
display(df_limpio.head())

# Mostramos el resumen estadístico de las columnas numéricas
print("\n--- ESTADÍSTICAS DESCRIPTIVAS ---")
display(df_limpio.describe())
```

### Rango de precios por marca de teléfono 

```python
# Histograma interactivo de precios segmentado por marcas reconocidas

fig_precios_marcas = px.histogram(
    df_limpio, 
    x='price_usd', 
    color='phone_brand', # Segmentación por bloques de color.
    title='Distribución General de Precios de Celulares según su Marca (2024)',
    labels={'price_usd': 'Precio en USD', 'count': 'Cantidad de Modelos', 'phone_brand': 'Marca'},
    barmode='stack',     # Apila los bloques de color uno encima del otro
    nbins=30             # Agrupa los precios en 30 barras para que la distribución sea fluida
)

# Mostrar el gráfico interactivo en el notebook
fig_precios_marcas.show()
```

### ¿Cómo influye la memoria RAM en el precio final de los celulares?

```python
# Gráfico de dispersión interactivo para analizar RAM vs Precio

fig_scatter_ram = px.scatter(
    df_limpio, 
    x='ram', 
    y='price_usd', 
    color='phone_brand',                   # Colorea los puntos por marca para identificar competidores
    hover_data=['phone_model', 'storage'], # Te muestra el modelo y almacenamiento al pasar el mouse
    title='Análisis de Impacto: Memoria RAM vs Precio en USD (2024)',
    labels={'ram': 'Memoria RAM (GB)', 'price_usd': 'Precio (USD)', 'phone_brand': 'Marca'}
)

# Mostrar el gráfico interactivo
fig_scatter_ram.show()
```
